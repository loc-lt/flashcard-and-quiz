```python
from datetime import datetime
import pytz

def get_electricity_rate_type_from_timestamp(timestamp_str):
    """
    Trả về loại giờ điện từ timestamp:
    1: Giờ thấp điểm
    2: Giờ bình thường
    3: Giờ cao điểm

    Args:
        timestamp_str: Chuỗi timestamp dạng '2025-05-29 08:05:00+07'
    """
    # Parse timestamp string
    if isinstance(timestamp_str, str):
        # Xử lý timezone nếu có
        if '+' in timestamp_str:
            dt = datetime.fromisoformat(timestamp_str.replace('+07', '+07:00'))
        else:
            dt = datetime.fromisoformat(timestamp_str)
    else:
        dt = timestamp_str

    # Chuyển về timezone Việt Nam nếu cần
    if dt.tzinfo is None:
        vietnam_tz = pytz.timezone('Asia/Ho_Chi_Minh')
        dt = vietnam_tz.localize(dt)

    current_hour = dt.hour
    current_minute = dt.minute
    current_time_minutes = current_hour * 60 + current_minute
    weekday = dt.weekday()  # 0=Monday, 6=Sunday

    def time_to_minutes(h, m=0):
        return h * 60 + m

    # Giờ thấp điểm: 22h00 - 04h00
    if current_time_minutes >= time_to_minutes(22) or current_time_minutes < time_to_minutes(4):
        return 1

    # Chủ nhật (weekday = 6) không có giờ cao điểm
    if weekday == 6:
        return 2

    # Thứ 2 - Thứ 7: Giờ cao điểm 09h30-11h30, 17h00-20h00
    if weekday <= 5:  # Thứ 2 - Thứ 7
        if ((time_to_minutes(9, 30) <= current_time_minutes < time_to_minutes(11, 30)) or
            (time_to_minutes(17) <= current_time_minutes < time_to_minutes(20))):
            return 3

    return 2
```

```python
get_electricity_rate_type_from_timestamp('2023-07-13 17:00:01.000')
```
